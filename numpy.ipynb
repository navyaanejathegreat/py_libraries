```python
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print ("Zomato data analysis:")
print("Sales data shape:",sales_data.shape)
print("Sample data for first three restaurants:\n",sales_data[:3])
print("total sales per year",np.sum(sales_data[:,1:],axis=0))
print("Minimum sales per restaurant:",np.min(sales_data[:,1:],axis=1))
print("Average sales per year:",np.average(sales_data[:,1:],axis=1))
cumsum=np.cumsum(sales_data[:,1:],axis=1)
print("cumulative sum",np.cumsum(sales_data[:,1:],axis=1))



#------------plotting 
plt.figure(figsize=(8,6))
plt.plot(np.mean(cumsum,axis=0))
plt.title("average cumulative sales across the restaurants")
plt.xlabel("Years")
plt.ylabel("Sales")
plt.grid('true')
plt.show()
```

```python


# Define the input image path and output npy file name
image_path = "numpy_logo.png" # Replace with your image file name/path
output_npy_file = "numpy_logo.npy"

def convert_image_to_npy(img_path, output_path):
    # 1. Open the image using Pillow
    try:
        # Open the image and convert to RGB (optional, handles different color modes)
        img = Image.open(img_path).convert("RGB") 
        print(f"Image opened successfully. Size: {img.size}, Mode: {img.mode}")
    except FileNotFoundError:
        print(f"Error: The file '{img_path}' was not found.")
        return
    except Exception as e:
        print(f"An error occurred while opening the image: {e}")
        return

    # 2. Convert the PIL image object to a NumPy array
    img_np_array = np.array(img)
    print(f"Image converted to NumPy array. Shape: {img_np_array.shape}, Dtype: {img_np_array.dtype}")

    # 3. Save the NumPy array as a .npy file
    np.save(output_path, img_np_array)
    print(f"Successfully saved NumPy array to '{output_path}'")

# Run the function
if __name__ == "__main__":
    convert_image_to_npy(image_path, output_npy_file)

    # Optional: Verify by loading the npy file back
    if os.path.exists(output_npy_file):
        loaded_array = np.load(output_npy_file)
        print(f"\nVerification successful: Loaded array shape {loaded_array.shape}")
```
