```python
###      SORTING 1D AND 2D ARRAYS

unsorted=np.array([2,3,4,2,32,1,43,3])
print(f"Sorted array:{np.sort(unsorted)}")

arr_2d_unsorted=np.array([[1,7],[4,3],[23,5],[24,3]])
print(f"Sorted array by row:\n{np.sort(arr_2d_unsorted)}")
print(f"Sorted array by column:\n{np.sort(arr_2d_unsorted,axis=0)}")
```

```python
###      FILTERING  
numbers=np.arange(10)
even_numbers=numbers[numbers%2==0]
print(even_numbers)
###      FILTERING USING MASK
mask= numbers>5
print(numbers[mask])
```

```python
###        FANCY INDEXING VS NP.WHERE()
indices=[0,2,4]
print(unsorted[indices])

#(exactly like mask technique)
where_result=np.where(unsorted>5)
print(unsorted[where_result])

#more ways to use np.where
condition_array=np.where(numbers>5,numbers*2,numbers)
print(condition_array)

# we take an example of if and else statement
#if(condition){operation}else{operation}
# if(numbers>5):
#     {numbers*4}
# else:
#     {numbers}
# print(numbers)
```

```python
###        FANCY INDEXING VS NP.WHERE()
indices=[0,2,4]
print(unsorted[indices])

#(exactly like mask technique)
where_result=np.where(unsorted>5)
print(unsorted[where_result])

#more ways to use np.where
condition_array=np.where(numbers>5,numbers*2,numbers)
print(condition_array)

# we take an example of if and else statement
#if(condition){operation}else{operation}
# if(numbers>5):
#     {numbers*4}
# else:
#     {numbers}
# print(numbers)
condition_array=np.where(numbers>5,'true','false')
print(condition_array)
```

PRACTICE QUESTIONS

```python
sales_data = np.array([
    [1, 150000, 180000, 220000, 250000],  # Paradise Biryani
    [2, 120000, 140000, 160000, 190000],  # Beijing Bites
    [3, 200000, 230000, 260000, 300000],  # Pizza Hub
    [4, 180000, 210000, 240000, 270000],  # Burger Point
    [5, 160000, 185000, 205000, 230000]   # Chai Point
])

print ("Zomato data analysis:")
print("Sales data shape:",sales_data.shape)
print("Sample data for first three restaurants:\n",sales_data[:3])
print("total sales per year",np.sum(sales_data[:,1:],axis=0))
print("Minimum sales per restaurant:",np.min(sales_data[:,1:],axis=1))
print("Average sales per year:",np.average(sales_data[:,1:],axis=1))
```
