```python
#SERIES
#Series is a one-dimensional labeled array capable of holding any 
#data type (integers, strings, floating point numbers, Python objects, etc.)
# data can be many different things:a Python dict,an ndarray,a scalar value

s1=pd.Series(np.random.randint(5,10,7))
s=pd.Series(np.random.randn(5))
s
```

```python
#data can also be fetched from dictionaries
d={"b":0,"a":2,"c":3}
e=pd.Series(d)
print(e)
print(e.loc['b'])
print(e.loc['a'])
print(e.loc['c'])
print(e.loc[:'a'])
#very very important slicing concepts
```

```python
#============================DATAFRAMES
arr = np.array([[1, 2], [3, 4]])
df = pd.DataFrame(arr, columns=["A", "B"])
df
```

```python
s1 = pd.Series([1, 2, 3], index=["a", "b", "c"])
s2 = pd.Series([10, 20], index=["b", "c"])

df = pd.DataFrame({"col1": s1, "col2": s2})
df
```

```python
#When a particular set of columns is passed along with a dict of data, 
# the passed columns override the keys in the dict.
data = {
    "A": [1, 2],
    "B": [3, 4]
}
df = pd.DataFrame(data)
print(df)
df = pd.DataFrame(
    data,
    columns=["B", "A"]
)
df
```

```python
# Column selection, addition, deletion
df=pd.DataFrame({"one":[1,2,3,np.nan]})
print(df)

df["two"]=[1,2,3,4]
print(df)
df["three"] = df["one"] * df["two"]
print(df)
#DELETING AND POPPING COLUMNS
del df["two"]
three = df.pop("three")
df
```
